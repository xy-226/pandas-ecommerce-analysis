```python
df.info()

# 查看每一列的缺失值数量
# isna() 会判断每个单元格是否为空
# sum() 会统计每一列有多少个空值
# sort_values(ascending=False) 会按缺失值数量从多到少排序
df.isna().sum().sort_values(ascending=False)



# 查看数据的整体统计摘要
# include="all" 表示数字列、文本列都一起统计
# 对数字列：显示均值、最小值、最大值、分位数等
# 对文本列：显示唯一值数量、出现最多的值、出现次数等
df.describe(include="all")
```

```python
# 统一列名，方便后续写代码
# 原始列名里有空格、大小写混合，比如 "Customer ID"、"BI Status"、" MV"
# 这些列名写代码时容易出错，所以先统一处理
df.columns = (
    df.columns
      .str.strip()   # 去掉列名前后的空格
      .str.lower()   # 全部转成小写
      .str.replace(" ", "_")
      .str.replace("-", "_")
)

df.columns.tolist()  # 查看处理后的列名
```

```python
# 查看订单状态分布
# value_counts() 会统计每个状态出现了多少次
# dropna=False 表示把缺失值 NaN 也统计出来
df["status"].value_counts(dropna=False)
```

```python

# 查看 BI 业务状态分布
# 这个字段可能是数据集整理后的业务状态
# 比 status 更适合做“完成/取消/退款”等大类分析
df["bi_status"].value_counts(dropna=False)
```

```python
# 查看商品一级分类出现次数最多的前 20 个
# head(20) 表示只看前 20 个，避免输出太长
df["category_name_1"].value_counts(dropna=False).head(20)
```

```python
# 查看支付方式分布
# 可以知道用户主要使用哪些支付方式
df["payment_method"].value_counts(dropna=False)
```

```python
# 把 created_at 转换成真正的日期时间类型
# errors="coerce" 表示如果某些值无法转换，就变成 NaT
# NaT 可以理解成日期类型里的缺失值
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

# 把 working_date 转换成日期时间类型
df["working_date"] = pd.to_datetime(df["working_date"], errors="coerce")

# 把 customer_since 转换成日期时间类型
df["customer_since"] = pd.to_datetime(df["customer_since"], errors="coerce")
```

```python
# 查看日期字段转换后的类型和缺失情况
df[["created_at", "working_date", "customer_since"]].info()
```

```python
# 先把 mv 字段统一转成字符串
# 因为有些值可能本来就是数字，有些可能是文本
# 统一转字符串后，才能安全地使用 str.replace 等字符串方法

df["mv"] = (
    df["mv"]
      .astype(str)                         # 转成字符串
      .str.replace(",", "", regex=False)   # 去掉数字里的逗号，比如 "1,200" -> "1200"
      .str.strip()                         # 去掉前后空格
)
```

```python
# 把清洗后的 mv 转成数字
# errors="coerce" 表示无法转换的值会变成 NaN
df["mv"] = pd.to_numeric(df["mv"], errors="coerce")
```

```python
# 查看 mv 字段的统计信息
# 如果转换成功，这里会显示 mean、min、max 等数值统计
df["mv"].describe()
```

```python
numeric_cols = ["price", "qty_ordered", "grand_total", "discount_amount", "mv"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")  
    #遍历每一个数值字段，尝试转成数字，转不了的变成缺失值
```

```python
# 从 created_at 中提取年月
# 例如 2017-03-15 会变成 "2017-03"
# 后面做月度趋势分析会用到
df["year_month"] = df["created_at"].dt.to_period("M").astype(str)
```

```python
# 判断订单是否使用了折扣
# discount_amount > 0 返回 True 或 False
# True 表示有折扣，False 表示没有折扣
df["has_discount"] = df["discount_amount"] > 0
```

```python
# 计算每一行的销售额估算值
# price 是单价，qty_ordered 是购买数量
# line_sales = 单价 × 数量
df["line_sales"] = df["price"] * df["qty_ordered"]
```

```python
df[["price", "qty_ordered", "line_sales", "grand_total", "mv"]].head()
```

```python
df.loc[3, [
    "increment_id",
    "status",
    "bi_status",
    "price",
    "qty_ordered",
    "line_sales",
    "grand_total",
    "discount_amount",
    "mv"
]]
```

通过检查样例订单发现，`grand_total = price × qty_ordered - discount_amount`。  
因此，`mv` / `line_sales` 更接近折扣前商品金额，`grand_total` 更接近折扣后实际支付金额。  
后续分析中，如果关注商品销售规模，使用 `mv`；如果关注实际收入，使用 `grand_total`。

```python
# 计算整体业务概况
# len(df)：总行数，也就是订单明细数量
# increment_id.nunique()：唯一订单数
# customer_id.nunique()：唯一客户数
# qty_ordered.sum()：总购买件数
# grand_total.sum()：订单总金额字段求和
# mv.sum()：mv 字段求和，可能代表商品销售额或 GMV
# groupby("increment_id")：按订单编号分组
# 每个订单的 grand_total 求和后，再取平均，得到平均订单金额

overview = {
    "rows": len(df),
    "unique_orders": df["increment_id"].nunique(),
    "unique_customers": df["customer_id"].nunique(),
    "total_qty": df["qty_ordered"].sum(),
    "total_grand_total": df["grand_total"].sum(),
    "total_mv": df["mv"].sum(),
    "avg_grand_total_per_order": df.groupby("increment_id")["grand_total"].sum().mean()
}

overview
```

```python
# 按 year_month 分组，统计每个月的核心指标
# order_count：唯一订单数
# item_count：订单明细行数
# total_qty：总购买数量
# total_grand_total：grand_total 总额
# total_mv：mv 总额

monthly_sales = (
    df.groupby("year_month", as_index=False)
      .agg(
          order_count=("increment_id", "nunique"),
          item_count=("item_id", "count"),
          total_qty=("qty_ordered", "sum"),
          total_grand_total=("grand_total", "sum"),
          total_mv=("mv", "sum")
      )
      .sort_values("year_month")
)

monthly_sales.head()
```

```python
# 按商品一级分类分组
# 统计每个品类的订单数、明细行数、购买数量和销售额

category_summary = (
    df.groupby("category_name_1", as_index=False)
      .agg(
          order_count=("increment_id", "nunique"),
          item_count=("item_id", "count"),
          total_qty=("qty_ordered", "sum"),
          total_mv=("mv", "sum"),
          total_grand_total=("grand_total", "sum")
      )
      .sort_values("total_mv", ascending=False)
)

category_summary.head(10)
```

```python
# 按支付方式分组
# 统计每种支付方式的订单数、总销售额和平均订单金额

payment_summary = (
    df.groupby("payment_method", as_index=False)
      .agg(
          order_count=("increment_id", "nunique"),
          total_mv=("mv", "sum"),
          avg_order_value=("grand_total", "mean")
      )
      .sort_values("order_count", ascending=False)
)

payment_summary
```

```python
# 按是否有折扣分组
# has_discount 是 True / False
# True 表示该订单明细使用了折扣
# False 表示没有使用折扣

discount_summary = (
    df.groupby("has_discount", as_index=False)
      .agg(
          order_count=("increment_id", "nunique"),
          total_mv=("mv", "sum"),
          avg_grand_total=("grand_total", "mean"),
          avg_discount=("discount_amount", "mean")
      )
)

discount_summary
```

```python
# 按折扣金额从高到低排序
# 查看折扣金额最高的前 20 条订单明细
# 这可以帮助我们发现异常折扣或大额优惠订单

top_discount_orders = df.sort_values("discount_amount", ascending=False)[
    [
        "increment_id",
        "category_name_1",
        "price",
        "qty_ordered",
        "grand_total",
        "discount_amount",
        "mv"
    ]
].head(20)

top_discount_orders
```

```python
import matplotlib.pyplot as plt

# 创建画布，figsize 控制图的大小
plt.figure(figsize=(14, 5))

# 画折线图
# x 轴是月份
# y 轴是每月 total_mv
plt.plot(monthly_sales["year_month"], monthly_sales["total_mv"])

# 让 x 轴月份文字旋转 45 度，避免重叠
plt.xticks(rotation=45)

# 图表标题和坐标轴名称
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total MV")

# 自动调整布局，避免文字被遮挡
plt.tight_layout()

# 显示图表
plt.show()
```
